### Task 1.1

Load `penguins.csv` with pandas. Print its shape and its column names.

```python
import pandas as pd
# Load the Dataset
# TODO: print df.shape and the list of df.columns

df = pd.read_csv('penguins.csv')
print('shape (rows, columns):', df.shape)
print('columns:', list(df.columns))
```

### Task 2.1

Compute the mean, median and variance of all four numeric attributes and present them as a table.

```python
# TODO: build a DataFrame with columns 'mean', 'median', 'variance'
# for df's numeric columns, and store it in a variable called desc
# Select only columns that contain numeric data
numeric_cols = df.select_dtypes('number')

# Calculate descriptive statistics for each numeric column
desc = pd.DataFrame({

    # Calculate the mean (average) of each numeric column
    'mean': numeric_cols.mean(),

    # Calculate the median (middle value) of each numeric column
    'median': numeric_cols.median(),

    # Calculate the variance (spread of values) of each numeric column
    'variance': numeric_cols.var(),

})

# Round all calculated values to 3 decimal places
desc.round(3)
```

### Task 3.1

Produce a histogram of one numeric attribute of your choice, split by species.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
# TODO: pick one numeric column and draw its histogram with sns.histplot,
# using hue='species'. Give the plot a title, then plt.show()
# Create a histogram
sns.histplot(
    data=df,                    # Use the DataFrame named df
    x='bill_length_mm',      # Put flipper length on the X-axis
    hue='species',              # Use different colors for each penguin species
    element='step',             # Draw histogram bars as step/outline shapes
    bins=20                     # Divide the flipper-length range into 20 intervals
)

# Add a title to the histogram
plt.title('Histogram of Bill Length by Species (Penguins dataset)')

# Display the plot
plt.show()
```

### Task 3.2

Produce a scatter plot of flipper_length_mm vs. body_mass_g, coloured by species.

```python
# TODO: draw a scatter plot with sns.scatterplot (x=flipper_length_mm,
# y=body_mass_g, hue='species'), give it a title, then plt.show()
# Create a scatter plot
sns.scatterplot(
    data=df,                    # Use the DataFrame named df
    x='flipper_length_mm',         # Put bill length on the X-axis
    y='body_mass_g',          # Put bill depth on the Y-axis
    hue='species',              # Use different colors for each species
    style='species'             # Use different marker shapes for each species
)

# Add a title to the plot
plt.title('Scatter Plot: Flipper Length vs Body Mass (Penguins dataset)')

# Display the plot
plt.show()
```
