```python
# Setup cell — run this first. Do not modify.
import pandas as pd
import numpy as np
from scipy.spatial.distance import euclidean, cityblock, pdist, squareform

df = pd.read_csv("iris.csv")
print(df.shape)
df.head()
```

### Task 1.1 — Building the data matrix (10 marks)

Using `petal_length` and `petal_width`:
1. Drop rows with missing values in either of these two columns.
2. Take rows 10 to 14 (i.e. the 5 rows at index positions 9–13 after dropping missing values and resetting the index) as your 5 data objects.
3. Store the result as a DataFrame called `X_df` with these two columns only, and label the 5 rows `s1` to `s5`.
4. Also store the plain NumPy array version as `X`.
5. Print `X_df` and the shape of `X`.

```python
# TODO: build X_df (5 rows x 2 columns: petal_length, petal_width, indexed s1..s5) and X (NumPy array)

# Remove rows where petal.length or petal.width is missing
# reset_index() creates new row positions after removing missing rows
clean = df.dropna(
    subset=["petal.length", "petal.width"]
).reset_index(drop=True)

# Select rows at index positions 9 to 13 (rows 10 to 14)
# Keep only the two numeric attributes
X_df = clean.iloc[9:14][["petal.length", "petal.width"]].copy()

# Rename the columns to petal_length and petal_width (the names used in the task)
X_df.columns = ["petal_length", "petal_width"]

# Rename the 5 data objects as s1, s2, ..., s5
X_df.index = [f"s{i+1}" for i in range(5)]

# Convert the DataFrame into a plain NumPy array
X = X_df.values

# Print the data matrix and the shape of the array
# Shape = (number of objects, number of attributes)
print("Data matrix (5 objects x 2 attributes):")
print(X_df)
print("\nShape of X:", X.shape)
```

### Task 2.1 — Euclidean distance: manual vs scipy (30 marks)

1. Write a function `manual_euclidean(a, b)` that computes the Euclidean distance between two 1-D NumPy arrays directly from the formula (no `scipy`).
2. For every pair of the 5 objects in `X`, print the manual result next to the result from `scipy.spatial.distance.euclidean` for the same pair.
3. Print the largest absolute difference between the manual and scipy results across all pairs.

```python
# TODO: define manual_euclidean(a, b) and compare it against scipy.spatial.distance.euclidean for every pair in X

# Function to calculate Euclidean distance using the formula
def manual_euclidean(a, b):

    # Subtract matching values, square them, add them, then take the square root
    return np.sqrt(np.sum((a - b) ** 2))


# Number of data objects
n = X.shape[0]

# Store the largest difference between manual and scipy results
max_diff_euc = 0.0

# Print column headings
print(f"{'Pair':8}{'manual':>12}{'scipy':>12}")

# Compare every unique pair of data objects
for i in range(n):
    for j in range(i + 1, n):

        # Euclidean distance using our own function
        manual = manual_euclidean(X[i], X[j])

        # Euclidean distance using scipy
        library = euclidean(X[i], X[j])

        # Keep the largest difference found so far
        max_diff_euc = max(max_diff_euc, abs(manual - library))

        # Label such as s1-s2
        label = f"{X_df.index[i]}-{X_df.index[j]}"

        # Show the manual result next to the scipy result
        print(f"{label:8}{manual:12.6f}{library:12.6f}")

# Largest difference across all pairs (close to 0 means both agree)
print(f"\nLargest |manual - scipy| difference: {max_diff_euc:.2e}")
```

### Task 2.2 — Manhattan distance: manual vs scipy (30 marks)

1. Write a function `manual_manhattan(a, b)` that computes the Manhattan (city-block) distance between two 1-D NumPy arrays directly from the formula (no `scipy`).
2. For every pair of the 5 objects in `X`, print the manual result next to the result from `scipy.spatial.distance.cityblock` for the same pair.
3. Print the largest absolute difference between the manual and scipy results across all pairs.

```python
# TODO: define manual_manhattan(a, b) and compare it against scipy.spatial.distance.cityblock for every pair in X

# Function to calculate Manhattan (city-block) distance using the formula
def manual_manhattan(a, b):

    # Add the absolute differences between matching values
    return np.sum(np.abs(a - b))


# Store the largest difference between manual and scipy results
max_diff_man = 0.0

# Print column headings
print(f"{'Pair':8}{'manual':>12}{'scipy':>12}")

# Compare every unique pair of data objects
for i in range(n):
    for j in range(i + 1, n):

        # Manhattan distance using our own function
        manual = manual_manhattan(X[i], X[j])

        # Manhattan distance using scipy (cityblock)
        library = cityblock(X[i], X[j])

        # Keep the largest difference found so far
        max_diff_man = max(max_diff_man, abs(manual - library))

        # Label such as s1-s2
        label = f"{X_df.index[i]}-{X_df.index[j]}"

        # Show the manual result next to the scipy result
        print(f"{label:8}{manual:12.6f}{library:12.6f}")

# Largest difference across all pairs (close to 0 means both agree)
print(f"\nLargest |manual - scipy| difference: {max_diff_man:.2e}")
```

### Task 3.1 — Dissimilarity matrix (30 marks)

1. Using `scipy.spatial.distance.pdist` and `squareform`, build the full 5×5 Euclidean dissimilarity matrix for `X`.
2. Using the same approach, build the full 5×5 Manhattan dissimilarity matrix for `X`.
3. Print both matrices as labelled pandas DataFrames (rows and columns labelled `s1`–`s5`), rounded to 3 decimal places.
4. Print which pair of objects is closest and which pair is furthest apart under each metric.

```python
# TODO: build and print the Euclidean and Manhattan dissimilarity matrices for X, and report the closest/furthest pairs

# Build the full 5 x 5 dissimilarity matrices
# pdist() calculates all pairwise distances at once
# squareform() converts them into a square, symmetric matrix
euc_matrix = squareform(pdist(X, metric="euclidean"))
man_matrix = squareform(pdist(X, metric="cityblock"))

# Convert both matrices into labelled tables rounded to 3 decimal places
euc_df = pd.DataFrame(np.round(euc_matrix, 3),
                      index=X_df.index, columns=X_df.index)
man_df = pd.DataFrame(np.round(man_matrix, 3),
                      index=X_df.index, columns=X_df.index)

# Display both matrices
print("Euclidean dissimilarity matrix:")
print(euc_df)
print("\nManhattan dissimilarity matrix:")
print(man_df)


# Function to find the closest and furthest pair(s) in a distance matrix
def closest_furthest(matrix, labels):

    # Upper half only, because the matrix is symmetric
    iu = np.triu_indices_from(matrix, k=1)

    # All unique pairs with their distance
    pairs = [(labels[i], labels[j], matrix[i, j]) for i, j in zip(*iu)]

    # Smallest and largest distances
    smallest = min(p[2] for p in pairs)
    largest = max(p[2] for p in pairs)

    # np.isclose() keeps pairs that tie, despite tiny floating-point differences
    closest = [p for p in pairs if np.isclose(p[2], smallest)]
    furthest = [p for p in pairs if np.isclose(p[2], largest)]

    return closest, furthest


# Find closest and furthest pair(s) under each metric
euc_close, euc_far = closest_furthest(euc_matrix, list(X_df.index))
man_close, man_far = closest_furthest(man_matrix, list(X_df.index))

# Display the results
print("\nEuclidean — closest:", ", ".join(f"{a}-{b}" for a, b, _ in euc_close),
      f"({euc_close[0][2]:.3f})")
print("Euclidean — furthest:", ", ".join(f"{a}-{b}" for a, b, _ in euc_far),
      f"({euc_far[0][2]:.3f})")
print("Manhattan — closest:", ", ".join(f"{a}-{b}" for a, b, _ in man_close),
      f"({man_close[0][2]:.3f})")
print("Manhattan — furthest:", ", ".join(f"{a}-{b}" for a, b, _ in man_far),
      f"({man_far[0][2]:.3f})")
```
