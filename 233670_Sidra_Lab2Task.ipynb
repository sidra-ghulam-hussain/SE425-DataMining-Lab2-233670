# Part 6: Loops and Conditionals - Tasks

**Task 6.1** For each reading in `readings`, print `Reading <position>: <value> -> <label>` where the label is `critical` if the value is 40 or more, `warning` if it is 30 or more, otherwise `normal`. Number positions from 1.

```python
# Purpose: your answer to task 6.1
readings = [22.5, 24.0, 31.5, 28.0, 41.2, 26.5]

# TODO: write your code below
for position, value in enumerate(readings, start=1):
    # thresholds checked from highest to lowest so each value gets one label
    if value >= 40:
        label = "critical"
    elif value >= 30:
        label = "warning"
    else:
        label = "normal"
    print(f"Reading {position}: {value} -> {label}")
```

**Task 6.2** Find the first reading in `readings` above 35. Print `First reading above 35: <value> (position <n>)`. If none exists, print `No reading above 35`. Use `break` and the loop's `else`.

```python
# Purpose: your answer to task 6.2
readings = [22.5, 24.0, 31.5, 28.0, 41.2, 26.5]

# TODO: write your code below
for position, value in enumerate(readings, start=1):
    if value > 35:
        print(f"First reading above 35: {value} (position {position})")
        break
else:
    # runs only if the loop completed without hitting break
    print("No reading above 35")
```

# Part 7: List and Dictionary Comprehensions - Tasks

**Task 7.1** Convert `raw_readings` into a list of floats and print it. Then print only the readings above 25, and print a dictionary that maps `r1`, `r2`, ... to each reading.

```python
# Purpose: your answer to task 7.1
raw_readings = "22.5,24.0,31.5,28.0,41.2,26.5"

# TODO: write your code below
readings_list = [float(value) for value in raw_readings.split(",")]
print(readings_list)

high_readings = [value for value in readings_list if value > 25]
print(high_readings)

readings_dict = {f"r{i}": value for i, value in enumerate(readings_list, start=1)}
print(readings_dict)
```

**Task 7.2** Print the total revenue: the sum of `unit_prices[i] * units_sold[i]` over all items, using `sum()` with `zip()`.

# Part 8: Functions and Pythonic Style - Tasks

**Task 8.1** Write `summarise(values)` that returns three values: the mean (rounded to 2 decimals), the minimum and the maximum. Call it on `temperatures` and print the result.

```python
# Purpose: your answer to task 8.1
temperatures = [22.5, 24.0, 31.5, 28.0, 41.2, 26.5]

# TODO: write your code below
def summarise(values):
    # round only the mean so min/max stay exact readings from the data
    mean_value = round(sum(values) / len(values), 2)
    return mean_value, min(values), max(values)

print(summarise(temperatures))
```

```python
# Purpose: your answer to task 8.2
# no setup data needed

# TODO: write your code below
def exam_eligibility(attendance_percent, minimum=75):
    if attendance_percent >= minimum:
        return "eligible"
    return "not eligible"

print(exam_eligibility(82))
print(exam_eligibility(68))
print(exam_eligibility(75))
print(exam_eligibility(78, minimum=80))
```
